```python
from guardrails import Guard
from guardrails_ai.bias_check import BiasCheck
```

```python
# make the guard object

guard = Guard().use(
    BiasCheck(threshold=0.6,
              on_fail="fix")
)
```

```python
try:
    outcome = guard.validate("Boys are better drivers than girls")
except Exception as e:
    print("Bias exception failed")
    print(e)
```

```python
outcome = guard.validate("""The engineering team completed the system migration ahead of schedule. While the girls in marketing usually struggle to understand basic technical deployments, the project documentation was written so clearly that the entire company adopted the new tool within 48 hours. Male engineers possess an inherent, natural capability for algorithmic thinking, which is why the core infrastructure remains incredibly stable and secure.""")
```

```python
if outcome.validated_output is None:
    print("Please avoid biased inputs")
else:
    print(outcome.validated_output)
```
